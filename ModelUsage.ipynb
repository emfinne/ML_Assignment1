# Model Training and Result Interpretation

This is where we both train and use machine learning algorithms for prediction. The goal is, like all of this, to predict the sales price of the property.
The three models that will be used are:
- `LinearRegression`
- `DecisionTreeRegressor`
- `RandomForestRegressor`

All of these will be used out-of-the-box without any tuning.

After training, these will be applied to the test dataset and various metrics will be measured:
- RMSE (Root Mean Squared Error): Measures the size of prediction errors in the same units as the `SalePrice` column. Lower is better.
- R² (R-squared): **Not actually `R*R`**. Compares the model predictions with a simple mean-prediction baseline. Higher is generally better. R² = 1 is perfect, R² = 0 is no better than predicting the mean, and R² can be negative which means worse than predicting the mean. R² cannot be described as an accuracy percentage.

```python
import numpy as np
import pandas as pd

trainSetName = "train_set_prepped.csv"
testSetName = "test_set_prepped.csv"

trainDf_x = pd.read_csv(trainSetName)
trainDf_y = trainDf_x.pop("SalePrice")

testDf_x = pd.read_csv(testSetName)
testDf_y = testDf_x.pop("SalePrice")
```

After loading in the datasets, we can put them into our models.
The general pattern for training and testing models is:
- `model.fit(training_set, to_predict_column)` for training
- `model.predict(test_set)` for testing which excludes the column to predict

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

#Learn
linRegModel = LinearRegression().fit(trainDf_x, trainDf_y)
decTreeModel = DecisionTreeRegressor().fit(trainDf_x, trainDf_y)
randForestModel = RandomForestRegressor().fit(trainDf_x, trainDf_y)

#Predict
linRegPredict = linRegModel.predict(testDf_x)
decTreePredict = decTreeModel.predict(testDf_x)
randForestPredict = randForestModel.predict(testDf_x)
```

Afterwards, we evaluate all the models using RMSE and R².

```python
from sklearn.metrics import r2_score

def rmse(actual, predicted) -> float:
    return np.sqrt(np.mean((actual - predicted) ** 2))

#Compare predicted values to testDf_y values using RMSE
linRegRMSE = rmse(testDf_y, linRegPredict)
decTreeRMSE = rmse(testDf_y, decTreePredict)
randForestRMSE = rmse(testDf_y, randForestPredict)

#Compare predicted values to testDf_y values using R^2 (this is still not R*R)
linRegR = r2_score(testDf_y, linRegPredict)
decTreeR = r2_score(testDf_y, decTreePredict)
randForestR = r2_score(testDf_y, randForestPredict)

print(f"LinReg          | RMSE: {linRegRMSE:.4f}, R²: {linRegR:.4f}")
print(f"DecisionTree    | RMSE: {decTreeRMSE:.4f}, R²: {decTreeR:.4f}")
print(f"RandomForest    | RMSE: {randForestRMSE:.4f}, R²: {randForestR:.4f}")
```
